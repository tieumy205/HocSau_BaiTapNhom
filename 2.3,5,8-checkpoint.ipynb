```python
#Bai8
def NhapMang():
    """
    Input:
        Ban phim = 1 4 1 2 3 4 1 2 3 3 1 2 3 4 10 2
    Output:
        a = [1, 4, 1, 2, 3, 4, 1, 2, 3, 3, 1, 2, 3, 4, 10, 2]
    """
    a = list(map(int, input("Nhap mang: ").split()))
    
    return a
# NhapMang
```

```python
import random
def SinhNgauNhien(n, vmin=-10, vmax=10):
    a = []
    for o in range(n):
        x = random.randint(vmin, vmax)
        a.append(x)
    
    return a
    
# SinhNgauNhien
b = SinhNgauNhien(30, -10, 10)
print(b)

b = SinhNgauNhien(30, vmin=0)
print(b)

b = SinhNgauNhien(30, vmax=2)
print(b)
```

```python
def DemTongChanLe(a):
    """
    Input: a[] = [1, 4, 2, 0, -1, -7, 3, -6, 4, 9, -5, 7, 7, -7, -10, 1, -10, 9, -4, -6, 10, -2, -10, -8, -3, 2, -7, -10, 6, 0]
    Output: tong, sochan, sole
    """
    tong = 0
    sochan = 0
    sole = 0
    for x in a:
       tong = tong + x
       if x % 2 == 0:
           sochan = sochan + 1
       else:
           sole = sole + 1
    return tong, sochan, sole
def DayChanLe(a):
    achan = []
    ale = []
    for x in a:
        if x % 2 == 0:
            achan.append(x)
        else:
            ale.append(x)
    return achan, ale
#DayChanLe

# DemTongChanLe
a = [1, 4, 2, 0, -1, -7, 3, -6, 4, 9, -5, 7, 7, -7, -10, 1, -10, 9, -4, -6, 10, -2, -10, -8, -3, 2, -7, -10, 6, 0]

tong, sochan, sole = DemTongChanLe(a)
print(f"Tong = {tong}")
print(f"So chan = {sochan}")
print(f"So le = {sole}")

achan, ale = DayChanLe(a)
print(f'DayChanLe:\nChan = {achan}\nLe = {ale}')
```
