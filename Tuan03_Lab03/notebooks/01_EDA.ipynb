```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("train.csv")

print("Kích thước dữ liệu:", df.shape)
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

missing
```

```python
missing_percent = (df.isnull().sum() / len(df) * 100)
missing_percent = missing_percent[missing_percent > 0].sort_values(ascending=False)

missing_percent
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["SalePrice"], bins=30, kde=True)

plt.title("Phân phối giá nhà (SalePrice)")
plt.xlabel("SalePrice")
plt.ylabel("Số lượng")
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

columns = ["GrLivArea", "LotArea", "TotalBsmtSF", "YearBuilt", "GarageArea"]

for ax, col in zip(axes.flat, columns):
    sns.histplot(df[col], bins=30, kde=True, ax=ax)
    ax.set_title(f"Phân phối {col}")
    ax.set_xlabel(col)
    ax.set_ylabel("Số lượng")

axes.flat[-1].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 5, figsize=(18, 5))

columns = ["SalePrice", "GrLivArea", "LotArea", "TotalBsmtSF", "GarageArea"]

for ax, col in zip(axes, columns):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)
    ax.set_ylabel("")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

columns = ["GrLivArea", "TotalBsmtSF", "GarageArea"]

for ax, col in zip(axes, columns):
    sns.scatterplot(data=df, x=col, y="SalePrice", ax=ax)
    ax.set_title(f"{col} và SalePrice")
    ax.set_xlabel(col)
    ax.set_ylabel("SalePrice")

plt.tight_layout()
plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

saleprice_corr = corr["SalePrice"].sort_values(ascending=False)

saleprice_corr
```

```python
top_features = [
    "SalePrice",
    "OverallQual",
    "GrLivArea",
    "GarageCars",
    "GarageArea",
    "TotalBsmtSF",
    "1stFlrSF",
    "FullBath",
    "TotRmsAbvGrd",
    "YearBuilt",
    "YearRemodAdd"
]

plt.figure(figsize=(10, 8))

sns.heatmap(
    df[top_features].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Heatmap tương quan giữa các biến")
plt.show()
```
