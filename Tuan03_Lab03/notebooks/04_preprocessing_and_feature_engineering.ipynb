# 02 — Preprocessing và Feature Engineering
Quan sát dữ liệu qua Cleaner → Feature Engineering → Preprocessor.

```python
from pathlib import Path
import sys, numpy as np, pandas as pd
import matplotlib.pyplot as plt
cwd=Path.cwd(); ROOT=cwd.parent if cwd.name.lower()=="notebooks" else cwd
SRC_DIR=ROOT/"src"
if str(SRC_DIR) not in sys.path: sys.path.append(str(SRC_DIR))
DATA_DIR=ROOT/"data"; EXPS_DIR=ROOT/"exps"; MODELS_DIR=ROOT/"models"/"sklearn"; SUBMISSIONS_DIR=ROOT/"submissions"
print("ROOT:",ROOT)
print("train.csv:",(DATA_DIR/"train.csv").exists())
```

```python
from preprocessing import HouseDataCleaner, build_preprocessor
from feature_engineering import HouseFeatureEngineer
train=pd.read_csv(DATA_DIR/"train.csv"); X=train.drop(columns=["SalePrice"]); y=train["SalePrice"]
print("Original:",X.shape)
```

```python
cleaner=HouseDataCleaner(); X_clean=cleaner.fit_transform(X,y)
print("After cleaner:",X_clean.shape)
display(X_clean.head())
```

```python
fe=HouseFeatureEngineer(); X_fe=fe.fit_transform(X_clean,y)
new_features=sorted(set(X_fe.columns)-set(X_clean.columns))
print("After FE:",X_fe.shape)
print("New features:")
for c in new_features: print("-",c)
```

```python
preprocessor=build_preprocessor(); Xp=preprocessor.fit_transform(X_fe,np.log1p(y))
print("Before preprocessing:",X_fe.shape)
print("After preprocessing :",Xp.shape)
fn=preprocessor.get_feature_names_out(); print("Số feature:",len(fn))
print(*fn[:30],sep="\n")
```
